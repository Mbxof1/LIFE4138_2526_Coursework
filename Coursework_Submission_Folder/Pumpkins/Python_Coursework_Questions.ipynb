### Question 7
Analyze the life expectancy dataset (which can be loaded by running the cell below).

- Calculate the number of unique countries in the dataset.

- Calculate the country and year with the highest life expectancy.

- Calculate the year with the highest total expenditure.

- Provide your code using pandas.


````python
import seaborn as sns
life_expectancy = sns.load_dataset('healthexp')
````

````python
#Print number of unique countries.
print(life_expectancy["Country"].nunique(),"\n")

#Calculate country and year combination with highest life expectancy, print both individually.
sorted_life_expectancy = life_expectancy.sort_values(by="Life_Expectancy",ascending=False)
print(sorted_life_expectancy["Country"].head(1),"\n")
print(sorted_life_expectancy["Year"].head(1),"\n")

#Combine expenditure based on year, sort low - high and print highest year.
life_expectancy_cost = life_expectancy.groupby("Year", as_index=False).sum()
sorted_life_expectancy_cost = life_expectancy_cost.sort_values("Spending_USD")
print(life_expectancy_cost["Year"].tail(1))

````

````python
import matplotlib.pyplot as plt
import numpy as np
import plotly.express as ptx

#Scatter plot, x axis carats, y axis price.
plt.scatter(diamonds["carat"],diamonds["price"])
plt.show()

#Histogram, showing frequency of different pricing values.
plt.hist(diamonds["price"])
plt.show()

#Boxplot showing cuts on x axis, and price on y axis.
diamondbox = ptx.box(diamonds, x="cut", y="price")
diamondbox.update_xaxes(categoryorder="array", categoryarray=["Fair","Good","Very Good","Premium","Ideal"])


````

### Question 10
Write a function that will convert carat to grams and find the heaviest diamond.

1 carat = 0.2 grams

````python
#Read in the highest carat value, create a function to convert carats into grams, then use the function and print the output in a text string.
maxcarat = max(diamonds["carat"])
def carattogram(caratvalue):
    return(float(caratvalue)*0.2)
maxcaratgram = carattogram(maxcarat)
print("The heaviest diamond in the dataset is "f"{maxcaratgram}{"g"}")
````
